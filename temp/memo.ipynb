流れ：
資料をAIに導入する（.xmlなどのformatがサポートされていませんので、.txtに変更する必要があります）
promptを作成する
    promptの構成要素
        元のdefault setting | new version のdefault setting
        元の実行のsetting(ai に導入するのが許可されない) | 変更後の実行のsetting | 
AIにをその差分ファイルを基に設定ファイルごとに質問をする。

問題点：
１．一部のデータのindexが失敗したことがあるので、それに関連する作成した設定ファイルの質量が悪くなるかもしれない。
１．１boxでフォルダとしてアップロードするのではなく、ローカルにアップロードすると機能する可能性があります。
2. AIと通信するときにチャットボックスに大きなファイルを入れると、トークンが不足してエラーが発生する可能性があります
